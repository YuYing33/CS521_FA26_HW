```python
import torch
from syncode import Syncode

print("CUDA:", torch.cuda.is_available())
print("SynCode imported successfully!")
```

```python
GRAMMAR = r'''
start: "{" "\"operands\"" ":" "[" number "," number "]" "," "\"operator\"" ":" operator "}"

number: SIGNED_NUMBER
operator: "\"+\"" | "\"-\"" | "\"*\"" | "\"/\""

%import common.SIGNED_NUMBER
%import common.WS
%ignore WS
'''

print(GRAMMAR)
```

```python

from syncode import Syncode

syn_llm = Syncode(
    model="microsoft/phi-2",
    grammar=GRAMMAR,
    mode="grammar_strict",
    parse_output_only=True,
    max_new_tokens=100,
    device="cuda",
    do_sample=False
)

print("Phi-2 with SynCode initialized!")
```

```python
FEW_SHOT_PROMPT = """Extract the two operands and the operator from each math question.
Return only a JSON object with "operands" and "operator".
Preserve the order of the operands. Do not calculate the result.

Question: What is 12.5 plus 3.2?
Extraction: {"operands":[12.5,3.2],"operator":"+"}

Question: What is 91.0 minus 15.4?
Extraction: {"operands":[91.0,15.4],"operator":"-"}

Question: What is 7.5 multiplied by 4.0?
Extraction: {"operands":[7.5,4.0],"operator":"*"}

Question: What is 64.0 divided by 8.0?
Extraction: {"operands":[64.0,8.0],"operator":"/"}

"""

print(FEW_SHOT_PROMPT)
```

```python
question = "What is 327. multiplied by 11.0?"

prompt = (
    FEW_SHOT_PROMPT
    + "Question: " + question + "\n"
    + "Extraction: "
)

output = syn_llm.infer(prompt)[0]

print("Model output:")
print(repr(output))
```

```python
import json
import operator

OPERATIONS = {
    "+": operator.add,
    "-": operator.sub,
    "*": operator.mul,
    "/": operator.truediv
}

def run_syncode(textual_query):
    # Step 1: Construct prompt
    prompt = (
        FEW_SHOT_PROMPT
        + "Question: " + textual_query + "\n"
        + "Extraction: "
    )

    # Step 2: Use Phi-2 with SynCode
    output = syn_llm.infer(prompt)[0]

    # Step 3: Parse generated JSON
    # raw_decode handles extra trailing text/newlines
    decoder = json.JSONDecoder()
    data, _ = decoder.raw_decode(output.strip())

    # Step 4: Extract operands and operator
    operands = [float(v) for v in data["operands"]]
    op = data["operator"]

    # Step 5: Validate extraction
    if len(operands) != 2:
        raise ValueError("Expected exactly two operands")

    if op not in OPERATIONS:
        raise ValueError("Invalid operator")

    # Step 6: Perform calculation in Python
    result = float(
        OPERATIONS[op](operands[0], operands[1])
    )

    # Step 7: Return required tuple
    return (operands, op, result)

print("run_syncode() defined successfully!")
```

```python

test_queries = [
    "What is 327. multiplied by 11.0?",
    "What is 45.1 plus 23.54?",
    "What is 120.4 divided by 4.0?"
]

expected = [
    ([327.0, 11.0], "*", 3597.0),
    ([45.1, 23.54], "+", 68.64),
    ([120.4, 4.0], "/", 30.1)
]

for query, target in zip(test_queries, expected):
    result = run_syncode(query)

    print("Query:", query)
    print("Output:", result)

    operands_ok = all(
        abs(a - b) < 1e-6
        for a, b in zip(result[0], target[0])
    ) and len(result[0]) == 2

    operator_ok = result[1] == target[1]
    answer_ok = abs(result[2] - target[2]) < 1e-6

    passed = operands_ok and operator_ok and answer_ok

    print("Result:", "PASS" if passed else "FAIL")
    print("-" * 50)
```

```python
additional_tests = [
    ("What is 95.5 minus 32.2?", [95.5, 32.2], "-", 63.3),
    ("What is -12.5 plus 7.5?", [-12.5, 7.5], "+", -5.0),
    ("What is 3.75 times 2.0?", [3.75, 2.0], "*", 7.5),
    ("What is 99.9 divided by 3.0?", [99.9, 3.0], "/", 33.3),
    ("Subtract 15.5 from 80.0.", [80.0, 15.5], "-", 64.5),
    ("What is 5.0 multiplied by 0.25?", [5.0, 0.25], "*", 1.25),
    ("Add 12.0 and 8.5.", [12.0, 8.5], "+", 20.5),
    ("Divide 144.0 by 12.0.", [144.0, 12.0], "/", 12.0)
]

passed_count = 0

for query, expected_nums, expected_op, expected_ans in additional_tests:
    try:
        nums, op, ans = run_syncode(query)

        passed = (
            len(nums) == 2
            and all(abs(a-b) < 1e-5
                    for a, b in zip(nums, expected_nums))
            and op == expected_op
            and abs(ans - expected_ans) < 1e-5
        )

        if passed:
            passed_count += 1

        print("PASS" if passed else "FAIL", "|", query)
        if not passed:
            print("  Got:", (nums, op, ans))
            print("  Expected:",
                  (expected_nums, expected_op, expected_ans))

    except Exception as e:
        print("ERROR |", query)
        print(" ", type(e).__name__, str(e))

print(f"\nPassed: {passed_count}/{len(additional_tests)}")
```
